## Pivot Tables

The `pivot_table()` function is used to summarize, aggregate, and reshape data in a spreadsheet-style format. It groups unique values from your data columns into a two-dimensional grid, making it highly effective for identifying trends and generating quick data insights.

```python
print("Pivot Table: Resolution SLA Compliance % by Agent Group\n")
piv_agent_sla = pd.crosstab(df['Agent Group'], df['SLA For Resolution'], normalize='index') * 100
print(piv_agent_sla)

print("\nPivot Table: Mean Resolution Time (Hours) by Priority & Support Level\n")
piv_res_time = df.pivot_table(index='Priority', columns='Support Level', values='Resolution Time (Hours)', aggfunc='mean')
print(piv_res_time)
```

##Observation

Pivot table provide summarized insights from the dataset.


*   The first table calculates the 'Resolution SLA Compliance % by Agent Group'.
*   The second table calculates the 'Mean Resolution Time (Hours) by Priority & Support Level'.



##Groupby

The `groupby()` method used to splits data into distinct groups, applies a function to each group, and combines the results. This follows the classic Split-Apply-Combine data analysis strategy.

```python
print("\nGroupby: First Response Time (Mins) & SLA Compliance by Topic\n")
grp_topic = df.groupby('Topic').agg(
    Avg_First_Respo=('First Response Time (Mins)', 'mean'),
    First_Resp_SLA_Met_Rate=('SLA For first response', lambda x: (x == 'Met').mean() * 100),
    Avg_Resolution_Hours=('Resolution Time (Hours)', 'mean'),
    Resolution_SLA_Met_Rate=('SLA For Resolution', lambda x: (x == 'Met').mean() * 100))
print(grp_topic)
```

##Observation

Groupby method analyze data based on the '**Topic**' column. For each unique Topic, it calculates four aggregated metrics.


*   **Avg_First_Respo** : Average 'First Response Time (Mins)' for tickets within that topic.
*   **First_Resp_SLA_Met_Rate** : The percentage of tickets within each topic where the 'SLA For first response' was 'Met'.
*   **Avg_Resolution_Hours** : This calculates the average 'Resolution Time (Hours)' for tickets belonging to each topic.
*   **Resolution_SLA_Met_Rate** : The percentage of tickets within each topic where the 'SLA For Resolution' was 'Met'.

#**Insights and Recommendations**


## 1. SLA compliance is generally very high.

First response time and Resolution time SLA compliance is above **96%** overall incident SLA performance is strong.

Also Resolution SLA compliance **97%** for all agent groups, with IT Support having the highest compliance (**99.29%**) and Customer Service the lowest (**97.59%**). This indicates that overall incident resolution performance is strong.

## 2. First response time has a very strong relationship with resolution time.

The correlation between First Response Time and Resolution Time is **0.966**, showing a very strong positive relationship. This means tickets that take longer to receive an initial response also tend to require more time to resolve. Improving initial response speed could therefore have a significant impact on overall resolution performance.





## 3. Priority has a clear impact on resolution time.

**Critical** tickets have the shortest average resolution time, around **0.84-0.86** hours, while **Low-priority** tickets take around **3.83-3.85** hours. This indicates that ticket priority is strongly associated with how quickly incidents are resolved.

## 4. Agent-group performance differs for first-response SLA.

**Security** group has the highest number of tickets meeting the first-response SLA and the lowest number of tickets failing it. **Customer Service** group has comparatively more first-response SLA failures, indicating an opportunity to improve response efficiency in this group.

##5. Software Bugs show the best SLA performance among topics.

**Software Bug** tickets have the highest first-response SLA compliance** (97.55%)** and the highest resolution SLA compliance **(99.30%)**. In contrast, **General Inquiry** has the lowest first-response SLA compliance **(94.54%)**, suggesting that this topic may require attention.

##6. Country-wise ticket volume is uneven.

The analysis shows that **Oman** has the highest ticket volume, while **Saudi Arabia** has the lowest. This suggests that support demand is not evenly distributed across countries and resource allocation may need to reflect ticket volume.

##7. Agent interactions have little direct relationship with response or resolution time.

The correlation between Agent Interactions and First Response Time is only **-0.030**, while its correlation with Resolution Time is **-0.024**. These values are very close to zero, suggesting that the number of agent interactions alone does not strongly explain how quickly tickets are responded to or resolved.

##Recommendations

*   Reduce first response time, since it is strongly associated with overall resolution time.
*   Investigate Customer Service and General Inquiry tickets to identify reasons for lower SLA performance.
*   Maintain the practices used by IT Support and Security, which demonstrate strong SLA performance.
*   Allocate resources according to country-wise ticket volume, particularly where ticket demand is highest.
*   Monitor low-priority tickets because they have substantially longer average resolution times.
*   Perform further analysis of ticket complexity, agent workload, and escalation patterns to identify additional factors affecting SLA compliance.
